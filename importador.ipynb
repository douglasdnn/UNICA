```python
import pandas as pd
import sqlite3
import os

def converter_xls_para_csv(caminho_xls: str, caminho_csv: str):
    """
    Função principal que lê um arquivo .xls e o salva como .csv.
    """
    try:
        # Verifica se o arquivo de entrada realmente existe antes de prosseguir
        if not os.path.exists(caminho_xls):
            print(f"\n❌ Erro: Arquivo não encontrado!")
            print(f"   Verifique se o nome '{caminho_xls}' está correto e se o arquivo está na mesma pasta do script.")
            return

        print(f"\nIniciando conversão de '{caminho_xls}'...")

        # Lê a planilha .xls usando pandas
        dataframe_xls = pd.read_excel(caminho_xls)

        # Salva o conteúdo em um arquivo .csv
        # index=False: para não criar uma coluna de índice no CSV.
        # encoding='utf-8-sig': para garantir que acentos e 'ç' funcionem bem no Excel.
        dataframe_xls.to_csv(caminho_csv, index=False, encoding='utf-8-sig')

        print(f"✅ Arquivo salvo com sucesso como '{caminho_csv}'")

    except Exception as e:
        print(f"\n❌ Ocorreu um erro inesperado durante a conversão: {e}")


    """
    Lê um arquivo CSV, ignora a primeira linha e escreve o restante
    em um novo arquivo CSV.

    Args:
        caminho_entrada (str): O caminho do CSV original.
        caminho_saida (str): O caminho onde o novo CSV será salvo.
    """
    try:
        with open(caminho_entrada, 'r', encoding='utf-8') as f_entrada, \
             open(caminho_saida, 'w', encoding='utf-8', newline='') as f_saida:
            
            # 1. Lê e descarta a primeira linha do arquivo de entrada
            next(f_entrada)
            
            # 2. Copia todas as linhas restantes para o arquivo de saída
            for linha in f_entrada:
                f_saida.write(linha)
                
        print(f"✅ A primeira linha de '{caminho_entrada}' foi removida e o resultado foi salvo em '{caminho_saida}'.")

    except FileNotFoundError:
        print(f"❌ Erro: O arquivo '{caminho_entrada}' não foi encontrado.")
    except Exception as e:
        print(f"❌ Ocorreu um erro: {e}")

nome_arquivo_input =input("Qual é o nome do arquivo?")
comarca=input("Qual é o nome da comarca?")

# 2. Garante que o nome do arquivo de entrada termina com .xls
if not nome_arquivo_input .lower().endswith('.xls'):
    arquivo_xls = nome_arquivo_input  + '.xls'
else:
    arquivo_xls = nome_arquivo_input 

# 3. Cria o nome do arquivo de saída trocando a extensão para .csv
#    os.path.splitext() divide 'sarandi.xls' em ('sarandi', '.xls')
nome_base = os.path.splitext(arquivo_xls)[0]
arquivo_csv = nome_base + '.csv'

# 4. Chama a função para realizar a conversão
converter_xls_para_csv(arquivo_xls, arquivo_csv)


# Nome do banco SQLite
caminho_banco = f'{nome_arquivo_input}.db'
# Nome da tabela no SQLite
nome_tabela = f'{comarca}'
# Ler o arquivo CSV em um DataFrame
df = pd.read_csv(arquivo_csv, skiprows=1)
# Adicionar uma coluna 'id' baseada no número da linha (começando em 1)
df.insert(0, 'id', range(1, len(df) + 1))
# Conectar ao banco SQLite
conn = sqlite3.connect(caminho_banco)
# Escrever a tabela no banco, sobrescrevendo caso já exista (ou use if_exists='append' para adicionar)
df.to_sql(nome_tabela, conn, if_exists='replace', index=False)
conn.close()
print("Importação concluída com sucesso.")

conn = sqlite3.connect(f'{nome_arquivo_input}.db')
cursor = conn.cursor()
query=f'ALTER TABLE {comarca} ADD conteudo TEXT;'
cursor.execute(query)
conn.commit()
conn.close()
print("Coluna conteúdo adicionada.")
```
